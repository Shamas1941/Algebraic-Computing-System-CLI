# Git and GitHub

Git is a version control system created by Linus Torvalds. Build around few commands and concepts.

## Git vs. GitHub Explained

In a project with multiple people GitHub works as a central hub with u and your teammates able to push their work on it, allowing it to be easily reviewed and merged. GitHub is now owned by Microsoft and is an opensource platform

### Git Workflow

#### local

The local part refers to you computer. It is where you computer files and all your work are stored.

In your computer the folder where you are working on your project is called a Working Directory. This is where your write your code, create new files and modify existing files. And is considered the first phase of Git workflow.

After you are satisfied with your work and want to save your changes that when you move onto the next stage of Git workflow. In this phase you move all your work to the Stage. It is the middle ground of the workflow.

Then comes the Third phase when we tke the staged files and send them to the local repository. Local repository isa temporary area where files sit between working directory and repository safe.

Once you have review everything and you are confident in your work, you Commit it. Which means permanently saving your changes to your local repository as a recorded version of your project history.

#### Remote (GitHub)

The remote part refer to a cloud server where you push and upload your remote work. You usually refers to GitHub when talking about remote.

#### Repository

A repository is a place where all the versions of your files and their complete history are stored. In the case of a local repository, Its a specific folder on your computer. For a remote repository it lives on a Cloud Server like GitHub.

Inside this Repository Git automatically creates a few system files that tracks everything : Your changes, History, Commits and more. These files are managed entirely by Git itself and the whole system runs based on the data stored within it

### Summary

1. You work inside your working directory.
2. Once were done we Stage the changes.
3. Then Commit those staged changes to our local repository
4. Lastly if we want to share our code with others, Then we can push our local repository with a remote repository (Uploading on GitHub)

#### Difference between git checkout, git switch and git restore

The last concept to understand on branching is the difference between <b>git checkout</b>, <b>git switch</b> and <b>git restore</b>. Although these commands all deal with navigating your project history, changing branches, or undoing changes.

Historically, <b>git checkout</b> was used to do everything switching branches, restoring old files, and detaching HEAD. Because it did so many different jobs, it was confusing and easy to make mistakes with. To fix this, modern Git introduced <b>git switch</b> and <b>git restore</b> to split those responsibilities into clear, single purpose commands.

|Command|Primary Job|What it Does|
|-------|-----------|---------------|
|git switch|Branches only|Dedicated entirely to moving between branches or creating new ones. It manages your project's context safely without touching working files directly.|

## WorkFlow Management

### Git Restore

Suppose you are working on a project, writing alot of new code and modifying alot of several existing files. After sometime you realize the approach you took isnt working out at all but by then you have already made changes across 10 to 15 different files. Some with new code and some heavily modified, in this situation u want to discard everything and go back to the previous state

Now think about it would u really open those 10-15 files manually, remove all the new lines and try to restore the old code one by one. Would such a thing even be accurate or possible?

In this case the simpliest answer is No, Its not possible at all, you can try as hard you want, but there is no guarantee that everything will return to its previous state. and this is exectly where the <b>git restore</b> commands comes. The git restore command helps you revert any files or directory back to its previous state. That is to the state of the last commit. It is mainly used to undo local uncommited changes. or to remove changes that were added to the staging area using <b>git add</b>

Suppose in the one.txt file i write "New Feature", now suppose i realized the change i made in the one.txt was actually a mistake. The previous version was fine. And since this change isnt ready to be commited, i want to undo it and go back to the last commited state, to do that i will simply run 
* git restore one.txt

As soon as you run this command the file will be instantly reverted back to its previous commited state. To restore changes across an entire directory, instead of just a single file just use the directory name after git restore
* git restore myFolder

Lastly another way to use the git restore command is <b>git restore .</b> Since The dot simply means the current working directory you are standing in right now, It will simply restore all the changes from that directory downwards
* git restore .

Lastly if u have already staged some changes using <b>git add</b> you can still undo them using <b>git restore --staged</b> command, which can also be used for all of the above inputs
* git restore --staged

but this removes the files from the staging area and keeps the working directory unchanged

<b>Note:</b> As mentioned previously git restore is a sub command or mordern command of git checkout, meaning all the context you have seen here can also be done using git checkout although, I wouldnt recommend using git checkout in exchange for git restore

### Git Stash

Now imagine this you are working on a new branch developing a new feature, Its a long process and you have already finished half of it, but the work isnt in a state to be commited. Suddenly another developer messages you telling you there is a new update on another branch that you need to check out, Now this raises the question regarding how u can change your branch without losing your unfinished work, are you going to through away all your work using the git restore command?

Ofcourse not, there is no reason to go throught that much hassel when git gives us a smarter way to deal with it, using the <b>git stash</b> command. With git stash you will temporarily set aside your unfinished work, switch to another branch to do something else, than later bring all your changes back with a single comment.

Suppose in your one.txt file you write "Another Feature", and although you are currently on your main branch but u want to switch to your developement branch to review something. so you type <b>git checkout developement</b> or <b>git switch developement</b>
* git switch developement

But git throws an error saying:  
Your local changes to the following files would be overwritten by checkout: one.txt

Please commit your changes or stash them before you switch branches. 

Aborting

This means git is protecting your work by telling you to commit before switching, or else it would overwritten. But since your feature isnt ready to be commited this is exectly when you use <b>git stash</b>
* git stash

The moment u run this comment all your uncommited changes will disappear from your working directory, git has safely stored all your work in a temporary area called stash. This lets u freely switch branches without affecting your unfinished work. Now lets switch to the developement branch
* git switch developement

Now git will allow u to switch without any warnings and after your done u can easily switch back
* git switch main

Now if u look at your one.txt file u will realize your uncommited work isnt there, to bring it back u can use the <b>git stash pop</b> command to bring it back
* git stash pop

Now u will realize all your work is back, the pop command restores all you recently stashed work and removes it from your stashed list. Stash list refers to the fact that if u have stashed multiple time all you stashes are store on git in accending order (newest on top and oldest on bottom) using the LIFO (last in first out) way.

Now lets say you want to unstash your work but dosent want to remove them from your stash list, to do this you can use the <b>apply</b> command instead.
* git stash
* git stash apply

Now all your changes will be restored while also being safely stored in the stash list.

To see the git stash list you simply run the command <b>git stash list</b>
* git stash list

When you run this command u will see a list of all your saved stashes with an identifying mark (i.e stash@{0}, stash@{1}, stash@{2}, etc). which u can use to apply or pop specific stashes aswell using <b>git stash pop stash@{0}</b> or <b>git stash apply stash@{0}</b>

To understand this better lets look at our stash list, as we recently used stash apply, we will have a "stash@{0}" appear, we can remove this manually using the <b>git stash drop</b>
* git stash list
* git stash drop

Now when u recheck your stash list, u will see that the stash@{0} has been removed, and since we used git apply, all the changes are still in our file directory, to check it we can simply use <b>git status</b>

Now lets stash these changes again
* git stash

Now if we check the stash list the "stash@{0}" will appear again now lets check out the git pop and apply commands
* git stash pop stash@{0}

Now u will see the "stash@{0}" stored item will be poped out of the stash list
* git stash
* git stash apply stash@{0}

Now u will see the "stash@{0}" being restored in your working directory but not being removed from your stash list, and lastly 
* git stash drop stash@{0}

Now u will see git drop the "stash@{0}" item from the stashed list

### Git Rebase

Imagine you start working on a new feature, so you create a new branch called "Feature", by checking out from your main branch, you begain developing a new feature, making changes bit by bit and commiting your progress as you go. Meanwhile another developer adds more updates to the main branch for production and now you also want to include those updates from the main branch into the Feature branch, So what can you do?

There are several ways to handle this, one of them is merging main branch into your Feature branch, If you do that git will add a new merge commit and merge the main branch with your feature branch. However remember that merging always creates an additional commit. Some developers find this extra merge commit a bit messy, and if you continue merging your commit history will look quite clusstered, making it harder to follow.

In such cases a better solution is to use <b>git rebase</b>, when you use rebase the base (starting point) of you feature branch changes to the latest point of the branch rebased is applied on (ie your Feature base changes to the latest main commit version). Then all your feature command commits will be reapplied on top of them. This makes it so the commit history becomes much cleaner and more linear. Which u can easily verify using <b>git log</b>

Now lets start with an example, suppose u make a new branch, and u switch to it
* git switch -c feature

Now we go to the file directory, open one.txt and type "Adding Dark Mode Functionality", then ad"d and commit it.
* start .
* git add .
* git commit -m "Adding Dark Mode Functionality"

Now lets switch back to the main branch, and create a new file inside the working directory called "two.txt" and just type "Adding Dark Mode UI", then stage and commit it.
* git switch main
* start .
* git add .
* git commit -m "Adding Dark Mode UI"

Now our main branch has a new feature called "Adding Dark Mode UI" now we want out feature branch to have that same feature as well, so first we switch back to the feature branch
* git switch feature

Now our goal is to move the new feature of the main branch into the feature branch, but this time we do it using rebase. To perform a <b>git rebase</b> you need to be on the branch you want to perform the rebase and specify the branch which you want to add those changes
* git rebase main

As soon as u run this all your new feature of the main branch will be added to the feature branch smoothly, and here is what happens behind the scenes
1. git first identifies the most recent common commit shared between your current branch (feature) and the branch you were rebasing of (main)
2. Then all the features in your feature branch that comes after your common commit is temporarily set aside
3. Now git applies all the commits from the main branch to the new branch (feature)
4. Finally git reapplies all the temporarily set aside commits from the feature branch on top of those main commits one by one.
5. The resuly is a much cleaner history then you can ever get with merging

<b>Note:</b> Even tho git rebase is such a powerful tool it is still recommended to not use it on public repositories or branches where multiple developers are working together. If you have to use it, you must inform your team before hand. Otherwise, it can cause serious issues. As git rebase rewrites the existing commit history, even the commit IDs change, so if you are working with someone and you use git rebase then your rebased commits wont match their local coppies anymore and they wont be able to pull or sync normally.

So before using it you must fully understand where you are applying it, and how it could affect your colligues or collaborators. Buts, Its perfectly resonable to use git rebase on your own local or personal branches where just you are working, just avoid rewriting commits that already exists on a shared remote branch

## Git Overview & Internal Workline

### Git (Behind the scenes)

We use Git everyday <b>commiting</b>, <b>pushing</b> and <b>pulling</b> etc, But the problem is most developers dont fully understand git, most use it for tasks withotu looking at what git does on the background or behind the scenes and most of the time this causes them to lose track of various changes and unlitmately lose alot of time.

<b>Remember although Git might not be as important as a programing language, but it is crucial if you want to work with others. And if you just use codes from memorization without knowning what git is actually doing in the background it could become a really big problem</b>

Alright lets look at git from zero forget everything we have learned untill now, Git is simply a database, and the fundamental unit of that data base is the commit which you can think of as a record or a snapshot. Dont think that by a snapshot i am refering each commit will restore all your files, instead each files stored by git contains an Hash and if you make a commit without changing a file or a sub folder, git dosent make a new copy.

To better visualize this think of your main folder which you initialize as a root folder of a tree (Or the main tree), any files stored in this root folder is called a blob in git (think of blobs as an hash of that file). Meanwhile sub folders are smaller trees. When you commit, Git creates a new tree. For any unchanged files or subfolders, the new tree simply points to the existing blobs and sub-trees in the database. For any files that were changed, Git creates a new blob (with a new hash). On your hard drive, it doesn't necessarily save that whole new file from scratch. Instead, it saves space by storing it as a diff (just the changes or "deltas") compared to the older version.

Apart from the files, commit also stores the metadata (Who created it, when it was created, and the commited message) and lastly a pointer to the parent commit, a parent commit is the commit that came before. When you make a new commit git saves the state of your new project and links it back to where you were, This creates a chain with each commit pointing to a parent commit (previous) and the parents point to their parents all the way back to the first commit. And that first commit is considered the origin (starting) point of your history, Just remember that commits always point backwards towards the origin

The reality about commits, If everyone commtis one by one, its a linear path simple, but real developement is alot messier, where you branch of for some feature, you collegue branch of for some bug fix, now we have one parent commit sharing with multiple children commits (branches), then you merge. now you have a commit with 2 parents. This structure in its entirity is called the DAG (directed acyclic graph)

Think of DAG kinda like a family tree, the directed means relationships only go one way (children pointing to parents), acyclic means no loops (nobody can be their own grandparent, meaning you can create a cycle in history). While a graph just refers to nodes (commits) and connections (pointing to their parent commit). This graph is your project history, every branch, every merge, every dicision any developer ever made was captured and stored in this structure. This is what makes Git powerful

### Branches & Head

Branches are actually simpler than you can think, most of the people starting git think of branches as some heavy complex things, thinking its a whole seperate copy of the code base, which is completely wrong think of branches as a sticky note of a commit, all it does is contain one piece of information the hash of a commit. When you create a new branch let say (Feature), git just says that (Feature) points to this specific commit ID, and whenever the person want type <b>git switch Feature</b> switch to this specific commit ID. Thats it, Whenever you commit on a branch, git creates a new commit ID for that specific commit and moves the sticky note name (Features) to it.

What about main? Well main isnt something special either, it just by convention we all have agreed for it to be the main line of work, we can understand it better by remembering that in older version of git instead of main we had master as primary line of work, nothing changed just the name changed to main to keep it simpler

Alight, we we know about commits and branchs, but how does git keep track of you? The answer is while working in git, git tracks you by a marker called HEAD, its git way of tracking your location, But instead of pointing at a specific commit, HEAD points at a branch, then the branch points at a commit, and that commit is your current location (HEAD -> Branch -> Commit [Your Location])

Now here is apart alot might not get, what happens to head when you go back into history, ie moving to a previous commit ID. Now HEAD point directly to that commit, no branch in between which is known as the "Detached head state" now you when switching to commit id why we need to use <b>--detach</b> atribute, In this state you can still work and commit but no branch will be follwing along and when you switch away those commits are orphaned meaning no parents branch point to them, and eventually gits garbage collection will clean them up (usually around 30-90 days). So unless you create a new branch in the detached state mostlikely any commits done will be orphaned

### Checkout/switch, Reset and Revert

Git has 3 areas where your code can live, your working directory, the staging area and lasty your local repository. Simply when you edit a file, git noticies the changes but dosent care untill u put it into the staging area (You are basically telling git to keep track of this file, i want this in my next commit), and when you commit git takes everything from the staging area and commit. 

This recap on staging area was important because of 3 commands <b>checkout/switch</b>, <b>reset</b> and <b>revert</b>. They all seem to undo things but performs completely different operations, checkout or switch simply moves HEAD, and in a detach state no commits change and no branches move. Commit history is practically untouched.

Now lets talk about reset, reset basically moves a branch, when you are on main and type <b>git reset "commit ID"</b> you are basically telling git to move to that commit ID, the commits that were ahead still exists in the database but becomes orphaned (meaning no parent commit points to them). Reset has three modes and each one affects the key areas differently <b>git reset --soft</b> moves the branch, but the staging area and working directory remains unchanged, meaning your undone changes becomes staged and ready to be commit again. You usually use it when you made three commits and want to combine them into one. <b>git reset --mixed</b> It is the default atribute of reset, It moves the branch and staging area, but working directory remains same. You usually use it when you commitied something but to recommit it differently, or maybe you want to split it into multiple commits. lastly <b>git reset --hard</b> It resets everything, your files change, uncommited work gone from your disk. Only use it when you want to completely abondand work and start from fresh

Revert dosent change anything, what it basically does is create a new commit that undoes the changes of the old commit. typically used when you want to undo something thats already been commited

### Reflog

Git reflog is a safety net, which u use incase of emergency, when you either reset hard or rebase wrong, commits are gone. This is when you run <b>git reflog</b>, it shows everywhere HEAD (you) has pointed (been) recently, every checkout, commit and reset. Those lost commits, those deleted history, they are prolly still here just that git has hidden them, and <b>git reflog</b> just shows their hash. So find the hash (commit ID), create a branch pointing to it, Congrats you have recovered your work. (Remember this though, git reflog entries expire usually around 90 days for reachable commits and 30 days for unreachable commits)
