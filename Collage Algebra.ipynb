## Proportions

$$\frac{n1}{d1} = \frac{n2}{d2}$$

```python
# Cross Multiplication
def cross_multiplication(n1, d1, n2, d2):
    if n1 == 0:
        answer = d1 * n2 / d2
    elif n2 == 0:
        answer = d2 * n1 / d1
    elif d1 == 0:
        answer = n1 * d2 / n2
    elif d2 ==0:
        answer = d1 * n2 / n1
    else:
        answer = (d2 * n1) / (d1 * n2)
    
    print(f"The value of the unknown is: {answer:.2f}")

```
